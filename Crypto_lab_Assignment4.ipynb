```python
#Level 1 - Q5
#Determine the number of lines, words and characters in cryptology.txt.
file=open("cryptology.txt", "r")

lines=0
words=0
characters=0

for line in file:
    lines += 1
    words += len(line.split())
    characters += len(line)

file.close()

print("Number of lines:", lines)
print("Number of words:", words)
print("Number of characters:", characters)
```

```python
#Level 1 - Q6
#Read cryptology_no.txt and count uppercase letters, lowercase letters, digits, spaces and special characters.
file=open("cryptology_no.txt", "r")

uppercase=0
lowercase=0
digits=0
spaces=0
special=0

for line in file:
    for ch in line:
        if ch.isupper():
            uppercase += 1
        elif ch.islower():
            lowercase += 1
        elif ch.isdigit():
            digits += 1
        elif ch.isspace():
            spaces += 1
        else:
            special += 1

file.close()

print("Uppercase letters:", uppercase)
print("Lowercase letters:", lowercase)
print("Digits:", digits)
print("Spaces:", spaces)
print("Special characters:", special)
```

```python
#Level 1 - Q7
#7. Write a program that accepts a word from the user and searches for that word in cryptology.txt.
word=input("Enter word to search: ")

file=open("cryptology.txt", "r")

found=False

for line in file:
    if word.lower() in line.lower():
        print("Word found:", line.strip())
        found=True

file.close()

if not found:
    print("Word not found")
```

```python
#Regex Q1
#1. Write a Python program to read security_data.txt and use a regular expression to identify and display all IPv4-address-like patterns present in the file.
import re

file=open("security_data.txt", "r")
text=file.read()
file.close()

ip_addresses=re.findall(r'\b(?:\d{1,3}\.){3}\d{1,3}\b', text)

print("IPv4-address-like patterns:")

for ip in ip_addresses:
    print(ip)
```

```python
#Regex Q2
#2. Write a Python program to identify and display all email-address-like patterns present in security_data.txt.
import re

file=open("security_data.txt", "r")
text=file.read()
file.close()

emails=re.findall(
    r'\b[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,}\b',
    text
)

print("Email addresses:")

for email in emails:
    print(email)
```

```python
#Regex Q3
#3. Write a Python program to extract session IDs that follow the format SEC-YYYY-NNNN, where YYYY and NNNN represent four digits.
import re

file=open("security_data.txt", "r")
text=file.read()
file.close()

session_ids=re.findall(r'\bSEC-\d{4}-\d{4}\b', text)

print("Session IDs:")

for session in session_ids:
    print(session)
```

```python
#Regex Q4
#4. Write a Python program to identify all hexadecimal sequences containing the characters 0–9, a–f, or A–F.
import re

file=open("security_data.txt", "r")
text=file.read()
file.close()

hex_values=re.findall(
    r'\b(?=[0-9A-Fa-f]*[A-Fa-f])[0-9A-Fa-f]+\b',
    text
)

print("Hexadecimal sequences:")

for value in hex_values:
    print(value)
```

```python
#Regex Q5
#5. Write a Python program to read security_data.txt and display all lines containing the word failed, irrespective of uppercase or lowercase.
import re

file=open("security_data.txt", "r")

for line in file:
    if re.search(r'\bfailed\b', line, re.IGNORECASE):
        print(line.strip())

file.close()
```

```python
#Regex Q6
#6. Write a Python program to identify and extract the port number from statements of the form Port 443 connection established.
import re

file=open("security_data.txt", "r")
text=file.read()
file.close()

port=re.search(
    r'\bPort\s+(\d+)\s+connection established\b',
    text
)

if port:
    print("Port number:", port.group(1))
else:
    print("Port number not found")
```
